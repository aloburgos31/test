```python
num1= float(input("First number: "))
num2= float(input("Second number:"))

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b != 0:
        return a / b

    else:
        return "Error: division by zero"

operation = input("Choose an operation, either (A(Addition), S(Subtraction), M(Multiplication), or D(Division)): ")

if operation == "A" or operation == "a":
    Answer="The sum of ",num1, "and ",num2, "is",add(num1, num2)
elif operation == "S" or operation == "s":
    Answer="The difference of ",num1, "and ",num2, "is",subtract(num1, num2)
elif operation == "M" or operation == "m":
    Answer="The product of ",num1, "and ",num2, "is",multiply(num1, num2)
elif operation == "D" or operation == "d":
    if num2 != 0:
        Answer="The quotient of ",num1, "and ",num2, "is",divide(num1, num2)
    else:
        print("Error: division by zero")
else:
    print("Invalid choice")

print(Answer)

# What functions did you create in your calculator program?: I created separate functions for addition, subtraction, multiplication, and division.
# What parameters did your functions use?: Each function used two float numbers as parameters.
# What arguments were passed when the functions were called?: The inputs entered by the user were passed as arguments.
# How did your program use the returned value?: The program assigned the result by displaying the returned value.
# Why is it better to divide the calculator program into functions instead of writing everything in one long block of code?: Using functions made the program easier to understand, fix, and reuse.
```
